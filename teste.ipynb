```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid", context="talk")

# escolha a métrica
metric = "nmi"   # "acc", "nmi", "ari", "f1_macro", "f1_micro"

df = pd.read_csv("outputs/comm_collapse_results.csv")
df = df[df["status"] == "ok"].copy()

# garante ordem
df["collapse_level"] = df["collapse_level"].astype(int)
df = df.sort_values(["collapse_level", "model"])

# =========================
# 1) Heatmap com valor bruto
# =========================
pivot_metric = df.pivot(index="model", columns="collapse_level", values=metric)

plt.figure(figsize=(12, 7))
sns.heatmap(
    pivot_metric,
    annot=True,
    fmt=".3f",
    cmap="YlGnBu",
    linewidths=0.5,
    cbar_kws={"label": metric.upper()},
)
plt.title(f"Desempenho por modelo e nível de colapso ({metric})")
plt.xlabel("Nível de colapso")
plt.ylabel("Modelo")
plt.tight_layout()
plt.show()

# =========================
# 2) Heatmap de ranking
# 1 = melhor modelo em cada collapse_level
# =========================
rank_df = df.copy()
rank_df["rank"] = rank_df.groupby("collapse_level")[metric].rank(ascending=False, method="min")
pivot_rank = rank_df.pivot(index="model", columns="collapse_level", values="rank")

plt.figure(figsize=(12, 7))
sns.heatmap(
    pivot_rank,
    annot=True,
    fmt=".0f",
    cmap="RdYlGn_r",
    linewidths=0.5,
    cbar_kws={"label": "Ranking (1 = melhor)"},
)
plt.title(f"Ranking dos modelos por nível de colapso ({metric})")
plt.xlabel("Nível de colapso")
plt.ylabel("Modelo")
plt.tight_layout()
plt.show()

# =========================
# 3) Boxplot para resumo geral
# =========================
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x="model", y=metric, color="#9ecae1")
sns.stripplot(data=df, x="model", y=metric, color="black", alpha=0.5, size=5)
plt.title(f"Distribuição de {metric} por modelo")
plt.xlabel("Modelo")
plt.ylabel(metric.upper())
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
# Célula 1: setup
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid", context="talk")

CSV_PATH = "outputs/comm_collapse_results.csv"
metric = "ari"          # troque aqui: 'acc', 'nmi', 'ari', 'f1_macro', 'f1_micro'
exclude_models = []     # opcional: ex. ['UCODE']

df = pd.read_csv(CSV_PATH)
df = df[df["status"] == "ok"].copy()
df["collapse_level"] = df["collapse_level"].astype(int)

if exclude_models:
    df = df[~df["model"].isin(exclude_models)].copy()

summary = (
    df.groupby(["model", "collapse_level"], as_index=False)[
        ["acc", "nmi", "ari", "f1_macro", "f1_micro"]
    ]
    .mean()
)

model_order = (
    summary.groupby("model")[metric]
    .mean()
    .sort_values(ascending=False)
    .index
    .tolist()
)

summary.head()
```

```python
# Célula 2: line plot por modelo
plt.figure(figsize=(13, 7))

sns.lineplot(
    data=summary,
    x="collapse_level",
    y=metric,
    hue="model",
    hue_order=model_order,
    marker="o",
    linewidth=2.2
)

plt.title(f"{metric.upper()} vs. collapse_level")
plt.xlabel("Collapse level")
plt.ylabel(metric.upper())
plt.xticks(sorted(summary["collapse_level"].unique()))
plt.legend(title="Model", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python
# Célula 3: heatmap de perda relativa em relação ao nível 0
base = (
    summary[summary["collapse_level"] == 0][["model", metric]]
    .rename(columns={metric: "baseline"})
)

rel = summary.merge(base, on="model", how="left")
rel["delta_vs_lvl0"] = rel[metric] - rel["baseline"]

heat = (
    rel.pivot(index="model", columns="collapse_level", values="delta_vs_lvl0")
    .loc[model_order]
)

plt.figure(figsize=(11, 6))
sns.heatmap(
    heat,
    annot=True,
    fmt=".3f",
    cmap="RdYlGn",
    center=0,
    linewidths=0.5,
    cbar_kws={"label": f"{metric.upper()} - baseline (level 0)"}
)

plt.title(f"Perda relativa de {metric.upper()} vs. baseline")
plt.xlabel("Collapse level")
plt.ylabel("Model")
plt.tight_layout()
plt.show()
```

```python
# Célula 4: slope chart de ranking entre level 0 e level 6
extremes = summary[summary["collapse_level"].isin([0, 6])][["model", "collapse_level", metric]]
wide = extremes.pivot(index="model", columns="collapse_level", values=metric).dropna()

r0 = wide[0].rank(ascending=False, method="min").astype(int)
r6 = wide[6].rank(ascending=False, method="min").astype(int)

plot_df = pd.DataFrame({
    "model": wide.index,
    "score_0": wide[0].values,
    "score_6": wide[6].values,
    "rank_0": r0.values,
    "rank_6": r6.values,
})
plot_df["rank_change"] = plot_df["rank_0"] - plot_df["rank_6"]

plt.figure(figsize=(10, 8))

for _, row in plot_df.iterrows():
    color = "#2a9d8f" if row["rank_change"] > 0 else "#e76f51" if row["rank_change"] < 0 else "#7f8c8d"
    plt.plot([0, 1], [row["rank_0"], row["rank_6"]], color=color, linewidth=2.5, alpha=0.9)
    plt.scatter([0, 1], [row["rank_0"], row["rank_6"]], color=color, s=70, zorder=3)
    plt.text(-0.03, row["rank_0"], f"{row['model']} ({row['score_0']:.3f})", ha="right", va="center", fontsize=10)
    plt.text(1.03, row["rank_6"], f"{row['model']} ({row['score_6']:.3f})", ha="left", va="center", fontsize=10)

plt.xlim(-0.25, 1.25)
plt.ylim(len(plot_df) + 0.5, 0.5)
plt.xticks([0, 1], ["collapse=0", "collapse=6"])
plt.yticks(range(1, len(plot_df) + 1))
plt.ylabel("Ranking")
plt.title(f"Troca de ranking em {metric.upper()}: collapse=0 vs collapse=6")
plt.grid(axis="y", alpha=0.2)
plt.tight_layout()
plt.show()
```

```python
# Célula 5: desempenho inicial vs robustez
wide_all = summary.pivot(index="model", columns="collapse_level", values=metric).dropna()
robust = pd.DataFrame({
    "model": wide_all.index,
    "score_lvl0": wide_all[0].values,
    "score_lvl6": wide_all[6].values,
})
robust["retention"] = robust["score_lvl6"] / robust["score_lvl0"]

plt.figure(figsize=(9, 7))
sns.scatterplot(
    data=robust,
    x="score_lvl0",
    y="retention",
    s=140
)

for _, row in robust.iterrows():
    plt.text(row["score_lvl0"] + 0.003, row["retention"] + 0.003, row["model"], fontsize=10)

plt.axhline(robust["retention"].mean(), linestyle="--", color="gray", alpha=0.7)
plt.xlabel(f"{metric.upper()} em collapse=0")
plt.ylabel("Retenção em collapse=6 / collapse=0")
plt.title(f"Desempenho inicial vs robustez em {metric.upper()}")
plt.tight_layout()
plt.show()
```

```python
# Célula 1: setup
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid", context="talk")

CSV_PATH = "outputs/comm_collapse_results.csv"
metric = "nmi"          # troque para acc, nmi, ari, f1_macro, f1_micro
exclude_models = []     # opcional: ['UCODE']

df = pd.read_csv(CSV_PATH)
df = df[df["status"] == "ok"].copy()
df["collapse_level"] = df["collapse_level"].astype(int)

if exclude_models:
    df = df[~df["model"].isin(exclude_models)].copy()

summary = (
    df.groupby(["model", "collapse_level"], as_index=False)[
        ["acc", "nmi", "ari", "f1_macro", "f1_micro"]
    ]
    .mean()
)

levels = sorted(summary["collapse_level"].unique())
models = sorted(summary["model"].unique())

summary.head()
```

```python
# Célula 2: bump chart de ranking ao longo do colapso


rank_df = summary.copy()
rank_df["rank"] = (
    rank_df.groupby("collapse_level")[metric]
    .rank(ascending=False, method="min")
)

order0 = (
    rank_df[rank_df["collapse_level"] == levels[0]]
    .sort_values(metric, ascending=False)["model"]
    .tolist()
)

plt.figure(figsize=(11, 8))

palette = sns.color_palette("tab20", n_colors=len(order0))
color_map = dict(zip(order0, palette))

for model in order0:
    sub = rank_df[rank_df["model"] == model].sort_values("collapse_level")
    plt.plot(
        sub["collapse_level"],
        sub["rank"],
        marker="o",
        linewidth=2.5,
        color=color_map[model],
        alpha=0.95
    )
    plt.text(
        sub["collapse_level"].min() - 0.15,
        sub["rank"].iloc[0],
        model,
        ha="right",
        va="center",
        fontsize=10,
        color=color_map[model]
    )
    plt.text(
        sub["collapse_level"].max() + 0.15,
        sub["rank"].iloc[-1],
        model,
        ha="left",
        va="center",
        fontsize=10,
        color=color_map[model]
    )

plt.gca().invert_yaxis()
plt.xticks(levels)
plt.yticks(range(1, len(order0) + 1))
plt.xlim(min(levels) - 0.5, max(levels) + 0.5)
plt.xlabel("Collapse level")
plt.ylabel("Ranking")
plt.title(f"Ranking dos algoritmos ao longo do colapso ({metric.upper()})")
plt.grid(axis="y", alpha=0.2)
plt.tight_layout()
plt.show()
```

```python
# Célula 4: dot plot em small multiples por nível
plot_df = summary.copy()

g = sns.FacetGrid(
    plot_df,
    col="collapse_level",
    col_wrap=4,
    height=4,
    sharex=False,
    sharey=True
)

def dotplot(data, **kwargs):
    data = data.sort_values(metric, ascending=True)
    ax = plt.gca()
    ax.hlines(y=data["model"], xmin=0, xmax=data[metric], color="#d9d9d9", linewidth=1.5)
    ax.scatter(data[metric], data["model"], s=70, color="#2a9d8f")

g.map_dataframe(dotplot)
g.set_titles("collapse={col_name}")
g.set_xlabels(metric.upper())
g.set_ylabels("Model")

for ax in g.axes.flat:
    ax.grid(axis="x", alpha=0.2)

g.fig.subplots_adjust(top=0.9)
g.fig.suptitle(f"Distribuição dos modelos por nível de colapso ({metric.upper()})")
plt.show()
```

```python
# Bloco 1: setup
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

from scipy.stats import spearmanr, kendalltau, linregress

sns.set_theme(style="whitegrid", context="talk")

# CSV_PATH = "outputs_old/heterophily_results.csv"
# CSV_PATH = "outputs/comm_collapse_results.csv"
CSV_PATH = "outputs/node_increase_results.csv"
metric = "nmi"  # acc, nmi, ari, f1_macro, f1_micro

# Ajuste conforme o CSV:
# comm_collapse_results.csv -> severity_col = "collapse_level"
# heterophily_results.csv   -> severity_col = "scenario_value"
# node_increase_results.csv -> severity_col = "scenario_value"
severity_col = "scenario_value"

df = pd.read_csv(CSV_PATH)
df = df[df["status"] == "ok"].copy()
df[severity_col] = pd.to_numeric(df[severity_col], errors="coerce")

summary = (
    df.groupby(["model", severity_col], as_index=False)[
        ["acc", "nmi", "ari", "f1_macro", "f1_micro"]
    ]
    .mean()
)

levels = sorted(summary[severity_col].dropna().unique())
model_order = (
    summary.groupby("model")[metric]
    .mean()
    .sort_values(ascending=False)
    .index
    .tolist()
)

summary.head()
```

```python
# Bloco 2: "average model tendency"
# Todas as curvas em cinza + média + IC aproximado de 95%

mean_df = (
    summary.groupby(severity_col)[metric]
    .agg(["mean", "std", "count"])
    .reset_index()
)
mean_df["se"] = mean_df["std"] / np.sqrt(mean_df["count"])
mean_df["ci95"] = 1.96 * mean_df["se"]

plt.figure(figsize=(11, 7))

for model in model_order:
    sub = summary[summary["model"] == model].sort_values(severity_col)
    plt.plot(
        sub[severity_col],
        sub[metric],
        color="lightgray",
        alpha=0.55,
        linewidth=1.7
    )

plt.plot(
    mean_df[severity_col],
    mean_df["mean"],
    color="#d94841",
    marker="o",
    linewidth=3,
    label="Mean across models"
)

plt.fill_between(
    mean_df[severity_col],
    mean_df["mean"] - mean_df["ci95"],
    mean_df["mean"] + mean_df["ci95"],
    color="#d94841",
    alpha=0.18,
    label="Approx. 95% CI"
)

plt.xticks(levels)
plt.xlabel(severity_col)
plt.ylabel(metric.upper())
plt.title(f"Average model tendency in {metric.upper()}")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Bloco 3: tendência média normalizada pelo baseline de cada modelo
# Bom para mostrar robustez relativa, não só valor absoluto

base = (
    summary.sort_values(severity_col)
    .groupby("model", as_index=False)
    .first()[["model", metric]]
    .rename(columns={metric: "baseline"})
)

norm_df = summary.merge(base, on="model", how="left")
norm_df["relative_score"] = norm_df[metric] / norm_df["baseline"]
norm_df["delta_score"] = norm_df[metric] - norm_df["baseline"]

mean_norm = (
    norm_df.groupby(severity_col)["relative_score"]
    .agg(["mean", "std", "count"])
    .reset_index()
)
mean_norm["se"] = mean_norm["std"] / np.sqrt(mean_norm["count"])
mean_norm["ci95"] = 1.96 * mean_norm["se"]

plt.figure(figsize=(11, 7))

for model in model_order:
    sub = norm_df[norm_df["model"] == model].sort_values(severity_col)
    plt.plot(
        sub[severity_col],
        sub["relative_score"],
        color="lightgray",
        alpha=0.5,
        linewidth=1.5
    )

plt.plot(
    mean_norm[severity_col],
    mean_norm["mean"],
    color="#2a9d8f",
    marker="o",
    linewidth=3,
    label="Mean relative score"
)

plt.fill_between(
    mean_norm[severity_col],
    mean_norm["mean"] - mean_norm["ci95"],
    mean_norm["mean"] + mean_norm["ci95"],
    color="#2a9d8f",
    alpha=0.18,
    label="Approx. 95% CI"
)

plt.axhline(1.0, linestyle="--", color="gray", alpha=0.7)
plt.xticks(levels)
plt.xlabel(severity_col)
plt.ylabel(f"{metric.upper()} / baseline")
plt.title(f"Relative robustness trend in {metric.upper()}")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Bloco 4: slope por modelo
# Quanto mais negativo, mais o modelo degrada com a severidade

rows = []
for model, sub in summary.groupby("model"):
    sub = sub.sort_values(severity_col)
    x = sub[severity_col].to_numpy(dtype=float)
    y = sub[metric].to_numpy(dtype=float)
    fit = linregress(x, y)
    rows.append({
        "model": model,
        "slope": fit.slope,
        "intercept": fit.intercept,
        "rvalue": fit.rvalue,
        "r2": fit.rvalue ** 2,
        "pvalue": fit.pvalue,
    })

slope_df = pd.DataFrame(rows).sort_values("slope", ascending=True)

plt.figure(figsize=(10, 7))
sns.barplot(data=slope_df, y="model", x="slope", color="#577590")
plt.axvline(0, linestyle="--", color="black", alpha=0.7)
plt.xlabel(f"Slope of {metric.upper()} vs {severity_col}")
plt.ylabel("Model")
plt.title(f"Sensitivity of each model to {severity_col}")
plt.tight_layout()
plt.show()

slope_df
```

```python
# Bloco 5: desempenho inicial vs retenção final
# Separa modelo "forte no começo" de modelo "robusto até o fim"

wide = summary.pivot(index="model", columns=severity_col, values=metric)
wide = wide.loc[model_order].dropna(axis=0)

first_level = min(wide.columns)
last_level = max(wide.columns)

robust_df = pd.DataFrame({
    "model": wide.index,
    "initial_score": wide[first_level].values,
    "final_score": wide[last_level].values,
})
robust_df["retention"] = robust_df["final_score"] / robust_df["initial_score"]
robust_df["drop_abs"] = robust_df["initial_score"] - robust_df["final_score"]

plt.figure(figsize=(9, 7))
sns.scatterplot(
    data=robust_df,
    x="initial_score",
    y="retention",
    s=140,
    color="#bc4749"
)

for _, row in robust_df.iterrows():
    plt.text(row["initial_score"] + 0.003, row["retention"] + 0.003, row["model"], fontsize=10)

plt.axhline(robust_df["retention"].mean(), linestyle="--", color="gray", alpha=0.7)
plt.xlabel(f"{metric.upper()} at first level")
plt.ylabel("Retention = final / initial")
plt.title(f"Initial performance vs robustness in {metric.upper()}")
plt.tight_layout()
plt.show()

robust_df.sort_values("retention", ascending=False)
```

```python
# Bloco 6: correlação de Spearman e Kendall por modelo
# Mede se a métrica cai monotonicamente com a severidade

corr_rows = []
for model, sub in summary.groupby("model"):
    sub = sub.sort_values(severity_col)
    x = sub[severity_col].to_numpy(dtype=float)
    y = sub[metric].to_numpy(dtype=float)

    rho, p_rho = spearmanr(x, y)
    tau, p_tau = kendalltau(x, y)

    corr_rows.append({
        "model": model,
        "spearman_rho": rho,
        "spearman_p": p_rho,
        "kendall_tau": tau,
        "kendall_p": p_tau,
    })

corr_df = pd.DataFrame(corr_rows).sort_values("spearman_rho")

plt.figure(figsize=(10, 7))
sns.barplot(data=corr_df, y="model", x="spearman_rho", color="#6a994e")
plt.axvline(0, linestyle="--", color="black", alpha=0.7)
plt.xlabel(f"Spearman rho: {severity_col} vs {metric.upper()}")
plt.ylabel("Model")
plt.title(f"Monotonic trend by model in {metric.upper()}")
plt.tight_layout()
plt.show()

corr_df
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid", context="talk")

metric = "nmi"  # acc, nmi, ari, f1_macro, f1_micro

scenarios = [
    {
        "title": "Community Collapse",
        "path": "outputs/comm_collapse_results.csv",
        "severity_col": "collapse_level",
    },
    {
        "title": "Node Increase",
        "path": "outputs/node_increase_results.csv",
        "severity_col": "scenario_value",
    },
    {
        "title": "Heterophily",
        "path": "outputs/heterophily_results.csv",
        "severity_col": "scenario_value",
    },
]

def build_robustness_df(csv_path, severity_col, metric):
    df = pd.read_csv(csv_path)
    df = df[df["status"] == "ok"].copy()
    df[severity_col] = pd.to_numeric(df[severity_col], errors="coerce")

    summary = (
        df.groupby(["model", severity_col], as_index=False)[
            ["acc", "nmi", "ari", "f1_macro", "f1_micro"]
        ]
        .mean()
    )

    wide = summary.pivot(index="model", columns=severity_col, values=metric).dropna(axis=0)
    wide = wide.reindex(sorted(wide.columns), axis=1)

    first_level = wide.columns.min()
    last_level = wide.columns.max()

    robust_df = pd.DataFrame({
        "model": wide.index,
        "initial_score": wide[first_level].values,
        "final_score": wide[last_level].values,
    })

    robust_df["retention"] = robust_df["final_score"] / robust_df["initial_score"]
    return robust_df

fig, axes = plt.subplots(1, 3, figsize=(18, 6), sharey=True)

for ax, spec in zip(axes, scenarios):
    robust_df = build_robustness_df(spec["path"], spec["severity_col"], metric)

    sns.scatterplot(
        data=robust_df,
        x="initial_score",
        y="retention",
        s=120,
        color="#2a9d8f",
        ax=ax
    )

    for _, row in robust_df.iterrows():
        ax.text(
            row["initial_score"] + 0.002,
            row["retention"] + 0.002,
            row["model"],
            fontsize=8
        )

    ax.axvline(robust_df["initial_score"].median(), linestyle="--", color="gray", alpha=0.7)
    ax.axhline(robust_df["retention"].median(), linestyle="--", color="gray", alpha=0.7)

    ax.set_title(spec["title"])
    ax.set_xlabel(f"{metric.upper()} at first level")
    ax.set_ylabel("Retention = final / initial")

plt.suptitle(f"Initial performance vs robustness across scenarios ({metric.upper()})", y=1.02)
plt.tight_layout()
plt.show()
```

```python
from __future__ import annotations

import argparse
import math
from collections import OrderedDict
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns


sns.set_theme(style="whitegrid", context="talk")


CLASS_TO_MODELS = OrderedDict(
    {
        "Reformulation Model": ["CoANE", "CoANE_topk", "DSAGC", "SANECE"],
        "KL-Div Model": ["DGC"],
        "Topological Model": ["SOLI", "NS4GC"],
        "Optimization Model": ["DGCluster", "UCODE", "EGC_VGAE"],
    }
)

DISPLAY_NAME = {
    "CoANE": "CoANE",
    "CoANE_topk": "CoANE top-k",
    "DGC": "DGC-EFR",
    "DSAGC": "DSAGC",
    "DGCluster": "DGCluster",
    "EGC_VGAE": "EG-VGAE",
    "NS4GC": "NS4GC",
    "SANECE": "SANECE",
    "SOLI": "SOLI",
    "UCODE": "UCoDe",
}

MODEL_TO_CLASS = {
    model: taxonomy_class
    for taxonomy_class, models in CLASS_TO_MODELS.items()
    for model in models
}

METRIC_LABEL = {
    "acc": "ACC",
    "nmi": "NMI",
    "ari": "ARI",
    "f1_macro": "F1 Macro",
    "f1_micro": "F1 Micro",
}

SCENARIO_LABEL = {
    "heterophily": "Heterophily",
    "edges": "Edges",
    "comm_collapse": "Collapse level",
}


def choose_x_column(df: pd.DataFrame) -> str:
    if "collapse_level" in df.columns and df["collapse_level"].notna().any():
        return "collapse_level"
    if "scenario_value" in df.columns and df["scenario_value"].notna().any():
        return "scenario_value"
    if "n_samples" in df.columns and df["n_samples"].notna().any():
        return "n_samples"
    raise ValueError("Could not infer the x-axis column from the CSV.")


def build_taxonomy_frame(csv_path: Path) -> pd.DataFrame:
    df = pd.read_csv(csv_path)
    df = df[df["status"] == "ok"].copy()
    df["taxonomy_class"] = df["model"].map(MODEL_TO_CLASS)
    df["display_model"] = df["model"].map(DISPLAY_NAME).fillna(df["model"])
    return df


def plot_one_csv(
    csv_path: Path,
    output_dir: Path,
    metrics: list[str],
    include_unclassified: bool,
) -> None:
    df = build_taxonomy_frame(csv_path)
    class_to_models = OrderedDict(CLASS_TO_MODELS)

    unclassified = sorted(df.loc[df["taxonomy_class"].isna(), "model"].dropna().unique())
    if unclassified:
        print(
            f"[{csv_path.name}] models outside the LaTeX taxonomy: "
            + ", ".join(unclassified)
        )

    if not include_unclassified:
        df = df[df["taxonomy_class"].notna()].copy()
    else:
        df["taxonomy_class"] = df["taxonomy_class"].fillna("Unclassified")
        if unclassified:
            class_to_models["Unclassified"] = unclassified

    if df.empty:
        print(f"[{csv_path.name}] no rows left after taxonomy filtering.")
        return

    x_col = choose_x_column(df)
    scenario = df["scenario"].dropna().iloc[0] if "scenario" in df.columns else ""
    x_label = SCENARIO_LABEL.get(scenario, x_col.replace("_", " ").title())

    class_order = [
        taxonomy_class
        for taxonomy_class in class_to_models
        if (df["taxonomy_class"] == taxonomy_class).any()
    ]

    ncols = 2
    nrows = math.ceil(len(class_order) / ncols)

    output_dir.mkdir(parents=True, exist_ok=True)

    for metric in metrics:
        if metric not in df.columns:
            print(f"[{csv_path.name}] metric '{metric}' not found, skipping.")
            continue

        plot_df = df[df[metric].notna()].copy()
        if plot_df.empty:
            print(f"[{csv_path.name}] metric '{metric}' has no valid rows, skipping.")
            continue

        fig, axes = plt.subplots(
            nrows=nrows,
            ncols=ncols,
            figsize=(7 * ncols, 4.8 * nrows),
            sharey=True,
        )
        axes = pd.Series(axes.ravel() if hasattr(axes, "ravel") else [axes])

        for ax, taxonomy_class in zip(axes, class_order):
            class_df = plot_df[plot_df["taxonomy_class"] == taxonomy_class].copy()
            for model_name in class_df["model"].drop_duplicates():
                model_df = class_df[class_df["model"] == model_name].sort_values(x_col)
                ax.plot(
                    model_df[x_col],
                    model_df[metric],
                    marker="o",
                    linewidth=2.2,
                    label=model_df["display_model"].iloc[0],
                )

            ax.set_title(taxonomy_class)
            ax.set_xlabel(x_label)
            ax.set_ylabel(METRIC_LABEL.get(metric, metric.upper()))
            ax.set_ylim(0.0, 1.0)
            ax.grid(True, linestyle="--", alpha=0.35)
            ax.legend(loc="best", fontsize=10, frameon=True)

        for ax in axes[len(class_order) :]:
            ax.axis("off")

        fig.suptitle(
            f"{csv_path.stem} | {METRIC_LABEL.get(metric, metric.upper())} by taxonomy",
            fontsize=18,
        )
        fig.tight_layout(rect=(0, 0, 1, 0.97))

        out_path = output_dir / f"{csv_path.stem}_{metric}_taxonomy.png"
        fig.savefig(out_path, dpi=220, bbox_inches="tight")
        plt.close(fig)
        print(f"saved: {out_path}")


def parse_args() -> argparse.Namespace:
    parser = argparse.ArgumentParser(
        description="Plot benchmark CSVs grouped by taxonomy class."
    )
    parser.add_argument(
        "--csv-dir",
        type=Path,
        default=Path("outputs/csvs"),
        help="Directory containing the benchmark CSV files.",
    )
    parser.add_argument(
        "--output-dir",
        type=Path,
        default=Path("outputs/taxonomy_plots"),
        help="Directory where the plots will be saved.",
    )
    parser.add_argument(
        "--metrics",
        nargs="+",
        default=["acc", "nmi", "ari", "f1_macro", "f1_micro"],
        help="Metrics to plot.",
    )
    parser.add_argument(
        "--include-unclassified",
        action="store_true",
        help="Also create an 'Unclassified' subplot for models missing from the LaTeX taxonomy.",
    )
    return parser.parse_args()


def main() -> None:
    args = parse_args()
    csv_paths = sorted(args.csv_dir.glob("*.csv"))
    if not csv_paths:
        raise FileNotFoundError(f"No CSV files found in {args.csv_dir}")

    for csv_path in csv_paths:
        plot_one_csv(
            csv_path=csv_path,
            output_dir=args.output_dir,
            metrics=args.metrics,
            include_unclassified=args.include_unclassified,
        )

main()
```
